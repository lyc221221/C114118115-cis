```python
import numpy as np
import pandas as pd

# 1. 基礎設定與參數
student_count = 20
departments = ["資工系", "電機系", "企管系", "資管系"]
rng = np.random.default_rng(42)

# 2. 產生學生基本資料與生活習慣數據
n = student_count
student_ids = np.arange(1, n + 1)
chosen_departments = rng.choice(departments, size=n)
sleep_values = rng.normal(6.7, 1.1, size=n).round(1)
screen_values = rng.normal(5.0, 1.5, size=n)
screen_values = screen_values.clip(1, 10).round(1)
study_values = rng.normal(2.5, 1.0, size=n)
study_values = study_values.clip(0.5, 6).round(1)
attendance_values = rng.normal(80.0, 10.0, size=n)
attendance_values = attendance_values.clip(60, 100).round().astype(int)

# 3. 建立原始 DataFrame
df = pd.DataFrame({
    "student_id": student_ids,
    "department": chosen_departments,
    "sleep_hours": sleep_values,
    "screen_time_hours": screen_values,
    "study_hours": study_values,
    "attendance_rate": attendance_values
})

# 4. 計算小考成績與加權
study_part = 3 * df["study_hours"]
sleep_part = 2 * df["sleep_hours"]
screen_part = -1.2 * df["screen_time_hours"]
attendance_part = 0.15 * df["attendance_rate"]
random_error = rng.normal(0, 5, size=n)

raw_scores = 50 + study_part
raw_scores = raw_scores + sleep_part
raw_scores = raw_scores + screen_part
raw_scores = raw_scores + attendance_part
raw_scores = raw_scores + random_error

df["quiz_score"] = raw_scores.clip(0, 100).round().astype("Int64")

# 5. 故意加入兩筆缺失值和一筆異常值 (依照圖片設定)
df.loc[[5, 17], "sleep_hours"] = np.nan
df.loc[8, "screen_time_hours"] = 15.0

# 6. 圖片中的特定檢查 display
print("兩筆睡眠缺失：")
display(df.loc[[5, 17], ["student_id", "sleep_hours"]])

print("一筆螢幕時間異常：")
display(df.loc[[8], ["student_id", "screen_time_hours"]])

# 7. 顯示全部資料表
print("\n完整資料表：")
pd.set_option('display.max_rows', 100)
selected_columns = [
    "student_id", "study_hours", "sleep_hours",
    "screen_time_hours", "attendance_rate", "quiz_score",
]
display(df[selected_columns])
```

```python
#【9問題】說明「資料列索引（Index）」與 student_id 為何差 1？
# 【答案】
# 1. 資料列索引 (Index)：Pandas 預設的資料列索引是從 0 開始編號（第 1 筆資料索引為 0）。
# 2. student_id：我們建立學號時是用 np.arange(1, n + 1)，設定學號從 1 開始編號（1, 2, 3...）。
# 3. 差異與對應：因為程式計數從 0 開始，學號從 1 開始，所以「索引值 + 1 = student_id」。
#    - 索引 5  對應 student_id  6（睡眠缺失 NaN）
#    - 索引 17 對應 student_id 18（睡眠缺失 NaN）
#    - 索引 8  對應 student_id  9（螢幕時間異常 15.0）
```
