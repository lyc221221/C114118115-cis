```python
# 一位假想學生的資料
study_hours = 2.5
sleep_hours = 7.0
screen_time_hours = 4.0
attendance_rate = 90

# 分別計算各項影響，再合計
base_score = 50
study_effect = 3 * study_hours
sleep_effect = 2 * sleep_hours
screen_effect = -1.2 * screen_time_hours
attendance_effect = 0.15 * attendance_rate

sample_score = base_score + study_effect
sample_score = sample_score + sleep_effect
sample_score = sample_score + screen_effect
sample_score = sample_score + attendance_effect

print("各項影響：", study_effect, sleep_effect, screen_effect, attendance_effect)
print("模擬成績：", round(sample_score, 1))
```

```python
# 依分數回傳描述
def describe_score(score):
    if score >= 80:
        return "表現良好"
    elif score >= 60:
        return "達到基本標準"
    else:
        return "需要加強"

print("82 分：", describe_score(82))
print("70 分：", describe_score(70))
print("55 分：", describe_score(55))
```
